```python
class Burger:
        def __init__(self, bun, patty, sauce):
                self.bun = bun
                self.patty = patty
                self.sauce = sauce

        def describe(self):
                print(f"This burger has a {self.bun} bun, {self.patty} patty and {self.sauce} sauce")
                
```

```python
myBurger = Burger("seasame","panner","mion")
myBurger.describe()
```

```python
from pydantic import BaseModel
class person(BaseModel):
    name:str
    age:int
```

```python
person1= person(name="AAA", age=24)
print(person1)
```

```python
class students(BaseModel):
    name:list
```

```python
stud = students(name={"aaa","bbb","ccc","ddd","eee"})
print(stud)
```

```python
class employees(BaseModel):
    emp:dict
```

```python
emp1=employees(emp={"name":"aaa","age":25})
print(emp1)
```

```python
class menu(BaseModel):
    det:list
    
```

```python
menuDet=menu(det=["idly","dosa","vadai","poori","pongal"])
print(menuDet)
```

```python
class ExtraC(BaseModel):
    classdet:set
```

```python
classdetails=ExtraC(classdet={"Chess","Table Tennis","Abacus"})
print(classdetails)
```

```python
import time

def soru():
    print("Starting the soru cooking process")
    time.sleep(2)
    print("soru is now cooking")
    time.sleep(5)
    print("soru is ready")

def kolambu():
    print ("Starting the kolambu cooking process")
    time.sleep(2)
    print("kolambu is simmering")
    time.sleep(5)
    print("kolambu is ready")

start=time.time()

soru=soru()
kolambu=kolambu()

end=time.time()

print(f"The total cooking time is:{end-start} seconds")

```

```python
import asyncio
import time

async def soru():
    print("The soru cooking process is starting")
    await asyncio.sleep(2)
    print("Soru is cooking now")
    await asyncio.sleep(5)
    print("soru is ready")

async def kolambu():
    print ("The kolambu cooking process is starting")
    await asyncio.sleep(2)
    print("Kolambu cooking is started")
    await asyncio.sleep(5)
    print("Kolambu is ready")

async def main():
    start=time.time()
    await asyncio.gather(soru(),kolambu())
    end=time.time()
    print(f"The time for cooking is:{end-start}")

await main()   # This is for ipynb file

# asyncio.run(main())   This is for normal python file

```
